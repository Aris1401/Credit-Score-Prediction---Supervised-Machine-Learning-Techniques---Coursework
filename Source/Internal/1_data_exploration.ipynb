# Visualizing the data

```python
import seaborn as sns
import matplotlib.pyplot as plt
import math
```

```python
num_cols = len(credit_score.columns)
cols_per_row = 3 
rows = math.ceil(num_cols / cols_per_row)

fig, axes = plt.subplots(rows, cols_per_row, figsize=(18, rows * 4))
axes = axes.flatten()

for i, col in enumerate(credit_score.columns):
    sns.histplot(credit_score[col].dropna(), ax=axes[i], kde=True, color='teal', bins=50)
    axes[i].set_title(f'Distribution of {col}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

| Category | Variables | Description | Relevance to Creditworthiness |
| :--- | :--- | :--- | :--- |
| **Demographics** | `A2`–`A4`, `A20` | Age, Gender, Education, Marital status | Establishes socio-economic stability. |
| **Employment & Income** | `A11`–`A14`, `A16`–`A18` | Total experience, Position, Tenure, Disposable income | Determines capacity to service debt. |
| **Loan Details** | `A10`, `A15` | Application duration, Loan amount | Used to measure leverage and debt burden. |
| **Credit & Payment History** | `A21`–`A45` | Arrears, Cheques returned, Debt ratio, Credit utilization | Directly measures historical repayment behavior (highest predictive power). |
